# Quick eda of tinyface data

```python
import sys
from zipfile import Path; sys.path.insert(0, '..')

def list_images(folder: str | Path, extensions=(".jpg", ".jpeg", ".png")) -> list[Path]:
    """Recursively list all image files in a folder."""
    folder = Path(folder)
    return sorted(p for ext in extensions for p in folder.rglob(f"*{ext}"))

from pathlib import Path

all_files = list_images('../data/tinyface_gallery_match')
grouped: dict[str, list[Path]] = {}

for file_path in all_files:
    # Extract identity_number from filename [identity_number]_[image_number].jpg
    stem = file_path.stem
    identity_number = stem.split('_')[0]
    
    if identity_number not in grouped:
        grouped[identity_number] = []
    grouped[identity_number].append(file_path)

num_images_per_identity = [len(images) for images in grouped.values()]
min_images = min(num_images_per_identity)
max_images = max(num_images_per_identity)
avg_images = sum(num_images_per_identity) / len(num_images_per_identity)

print(f"Min images per identity: {min_images}")
print(f"Max images per identity: {max_images}")
print(f"Average images per identity: {avg_images:.2f}")
```
